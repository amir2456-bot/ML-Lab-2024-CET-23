TASK 07 EXTRACT AN EMAIL DOMAIN (write a function part that return after@)

```python
def get_domain(email):
    """Return the domain portion of an email address."""
    return email.split("@")[-1]

print("Q7:", get_domain("user@domain.com"))
```

TASK 08 CHECK FOR WORD DOG(checking without case sensitivity)

```python
def contains_dog(text):
    """Return True if the text contains 'dog', ignoring case."""
    return "dog" in text.lower()

print(f"Q8: {contains_dog('Is there a dog here?')}")
print(f"Q8: {contains_dog('My DOG is big')}")
print(f"Q8: {contains_dog('Is there a cat here?')}")
```

TASK 09 COUNT OCCURENCES OF DOG (count matches regardless of letter case)

```python
def count_dog(text):
    """Count occurrences of 'dog', ignoring case."""
    return text.lower().count("dog")

print(f"Q9: {count_dog('This dog runs faster than the other dog dog')}")
print(f"Q9: {count_dog('Dog DOG dog')}")
```

TASK 10 SPEEDING TICKET (return the ticket category allowing on extra 5km on birthday)

```python
def caught_speeding(speed, is_birthday):
    """Choose a ticket category, with a 5 km/h birthday allowance."""
    allowance = 5 if is_birthday else 0

    if speed <= 60 + allowance:
        return "No Ticket"
    elif speed <= 80 + allowance:
        return "Small Ticket"
    return "Big Ticket"

print(f"Q10: {caught_speeding(81, True)}")
print(f"Q10: {caught_speeding(81, False)}")
print(f"Q10: {caught_speeding(65, True)}")
```
