TASK 03 (RANDOMNESS (A-E))

```python
np.random.seed(2026)
r = np.random.rand()
print('A: rand =', r)

rn = np.random.randn(25)
print('B: randn =', rn)

np.random.seed(42)
dice = np.random.randint(1, 7, size=10000)
faces, counts = np.unique(dice, return_counts=True)
for f, c in zip(faces, counts):
    print(f'Face {f}: {c}')

p_six = np.mean(dice == 6)
print('D: P(six) ~', p_six, ' (theory = 1/6 =', round(1/6, 4), ')')

pairs = dice.reshape(-1, 2)                      # 5000 pairs
double_sixes = np.sum((pairs[:, 0] == 6) & (pairs[:, 1] == 6))
print('E: double sixes =', double_sixes, 'out of', len(pairs), 'pairs')
```

TASK 05 (EXAM HALL SEATING PLAN (A-F))

```python
# A) Hall layout
hall = np.arange(1, 26).reshape(5, 5)
print('A:\n', hall)

# B) hall[2:, 1:]  -> rows 2 to end, columns 1 to end
print('B:\n', hall[2:, 1:])
# Invigilator points at the bottom-right 3x4 block: seats 12-15, 17-20, 22-25

# C) hall[3, 4]  -> single seat, row 3, column 4
print('C:', hall[3, 4])
# Invigilator points at seat 20
# D) hall[0:3, 1:2]  -> rows 0-2, column 1 only (stays 2D, shape 3x1)
print('D:\n', hall[0:3, 1:2])
# Invigilator points at seats 2, 7, 12 (second column, top three rows)

# E) hall[4] and hall[3:, ]
print('E1: hall[4]  =', hall[4])      # entire last row: seats 21-25
print('E2: hall[3:,]=\n', hall[3:, ]) # last two rows: seats 16-25

# F) Sum and std of all seats
print('F: sum =', hall.sum(), ' std =', hall.std())
```
